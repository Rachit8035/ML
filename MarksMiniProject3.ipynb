MINI PROJECT: Student Exam Score Analyzer
==============================================
 
You're given the exam scores of 6 students across 5 subjects
(Math, Science, English, History, Computer).
 
Using ONLY NumPy (no pandas), write code to:
 
1. Store the scores in a 2D NumPy array (6 rows = students, 5 cols = subjects)
2. Find each student's total and average score
3. Find each subject's average score (class average per subject)
4. Find the topper (student with highest total score)
5. Find the subject where the class performed weakest (lowest average)
6. Grade each student based on their average:
      >= 90 -> "A"
      >= 75 -> "B"
      >= 60 -> "C"
      else  -> "D"
7. Normalize all scores to a 0-1 scale (min-max normalization)
8. Print a clean report
 
Try writing this yourself first before scrolling to the solution below.
Data to use:
 
Students: Rachit, Amit, Priya, Neha, Karan, Sneha
Subjects: Math, Science, English, History, Computer
 
scores = [
    [85, 92, 78, 88, 95],
    
    [70, 65, 80, 72, 68],
    
    [95, 98, 92, 90, 97],
    
    [60, 55, 70, 65, 58],
    
    [88, 84, 79, 91, 86],
    
    [45, 50, 60, 40, 52],
]

```python
import numpy as np
student=np.array(["Rachit", "Amit", "Priya", "Neha", "Karan", "Sneha"])
subjects=np.array(["Math", "Science", "English", "History", "Computer"])
scores=np.array([
    [85, 92, 78, 88, 95],
    
    [70, 65, 80, 72, 68],
    
    [95, 98, 92, 90, 97],
    
    [60, 55, 70, 65, 58],
    
    [88, 84, 79, 91, 86],
    
    [45, 50, 60, 40, 52],
])

# Each student's total and average marks
totalMark=scores.sum(axis=1)
averageMarks=scores.mean(axis=1)
print(f"Total marks={totalMark}")
print(f"Avg marks={averageMarks}")

#Average score per subject
averageMarkSubject=scores.mean(axis=0)
print(f"Avg Sub marks={averageMarkSubject}")

#Topper->With highest total
topper_index=np.argmax(totalMark)
topper_name=student[topper_index]
print(f"Topper ={topper_name}")

#Lowest Subject
lowerSub_index=np.argmin(averageMarkSubject)
lowerSub_name=subjects[lowerSub_index]
print(f"Topper ={lowerSub_name}")


print("="*55)
print("EXAM RESULTS")
print("="*55)

for i in range(len(student)):
    print(f"{student[i]:<8}|Total:{totalMark[i]:>3}")

print("-"*55)
print(f"Topper:{student[topper_index]}")
print("-"*55)
print(f"Lowest subject:{subjects[lowerSub_index]}")
```
